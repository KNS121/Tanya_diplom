```python
import json
import psycopg2
from psycopg2 import OperationalError


def get_unique_values(config_path: str, column_name: str, table_name: str):
    """
    Получает уникальные значения из указанного столбца таблицы.
    :param config_path: Путь к JSON-файлу с конфигурацией подключения.
    :param column_name: Имя столбца.
    :param table_name: Имя таблицы.
    :return: Список уникальных значений.
    """
    # Загружаем конфигурацию
    try:
        with open(config_path, 'r') as file:
            config = json.load(file)
    except FileNotFoundError:
        print(f"Файл конфигурации {config_path} не найден!")
        exit(1)
    except json.JSONDecodeError:
        print(f"Ошибка в формате JSON в файле {config_path}!")
        exit(1)

    unique_values = []
    try:
        # Устанавливаем соединение с базой данных
        conn = psycopg2.connect(
            dbname=config['dbname'],
            user=config['user'],
            password=config['password'],
            host=config['host'],
            port=config['port']
        )
        # Создаем курсор для выполнения SQL-запросов
        cur = conn.cursor()

        # Выполняем SQL-запрос для выбора уникальных значений
        cur.execute(f'SELECT DISTINCT "{column_name}" FROM {table_name}')

        # Получаем результаты
        unique_values = [row[0] for row in cur.fetchall()]

    except OperationalError as e:
        print(f"Ошибка при подключении к базе данных: {e}")
    finally:
        # Закрываем курсор и соединение
        if cur:
            cur.close()
        if conn:
            conn.close()

    return unique_values
```

```python
if __name__ == "__main__":
    # Путь к конфигурационному файлу
    config_path = 'config_to_connection.json'

    # Имя столбца и таблицы
    column_name = 'your_column'  # Замените на имя вашего столбца
    table_name = 'experiment_full_table'  # Замените на имя вашей таблицы

    # Получаем уникальные значения
    unique_values = get_unique_values(config_path, "Material", "experiment_full_table")
    print("Уникальные значения из столбца:", unique_values)
```
