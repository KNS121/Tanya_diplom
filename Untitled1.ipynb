```python
Array_catalogs = get_cat_name.get_name_catalog(dir_name_with_data)
Array_catalogs.pop(0)
```

```python
df_good_materials = pd.read_csv('df_good_materials.csv')
```

```python
df_result_de_dt = pd.DataFrame(columns = ['Номер образца', 'de_dt'])
```

```python
df_result_de_dt
```

```python
for catalog in Array_catalogs:
    if catalog in df_good_materials['0'].values:
        path = os.path.join(dir_name_with_data, catalog, 'Test file 1', f'{catalog}.csv')

        df = Trying_to_read_csv.try_to_read_csv(path, 'cp1251', 'utf8')

        df_checking_material = df_lin_window[df_lin_window['Номер образца'] == catalog]

        if not df_checking_material.empty:
            start_time = df_checking_material['Начало'].astype(float).values[0]
            end_time = df_checking_material['Конец'].astype(float).values[0]

            df['Время, ч'] = df['Время, ч'].astype(float)

            df = df[(df['Время, ч'] >= start_time) & (df['Время, ч'] <= end_time)]

            print(f"Start time: {start_time}")
            print(f"End time: {end_time}")
            print(df)
            
            print('----------------------------------------------')
            
            de_dt = ((df['Деформация, %'].iloc[-1] - df['Деформация, %'].iloc[0]) /
                 (df['Время, ч'].iloc[-1] - df['Время, ч'].iloc[0]))
                
            row = {'Номер образца': catalog, 'de_dt': de_dt}
            
            df_result_de_dt = df_result_de_dt.append(row, ignore_index=True)
            
            
        else:
            print(f"No matching material found for catalog: {catalog}")
```

```python
df_result_de_dt
```

```python
merged_df = pd.merge(df_lin_window, df_result_de_dt, on='Номер образца', how='inner')
```
