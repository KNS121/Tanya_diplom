```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
```

```python
df_for_train = pd.read_csv("df_for_train.csv")
```

```python
df_for_train
```

```python
df_result = pd.DataFrame(columns = ["Номер образца", "A, 1 / [ сут * кг/см^2 ]", "n"])
```

```python
df_result
```

```python
for material in df_for_train["Номер образца"]:
    
    sample_data = df_for_train[df_for_train["Номер образца"] == material]
    
    x = np.array(sample_data["Нагрузка (MPa)"])*(10**(4))/((10**7))
    y = np.array(sample_data["de_dt"])/24
    
    X = np.log(x)
    Y = np.log(y)
    
    coefficients = np.polyfit(X, Y, 1)
    n = coefficients[0]
    B = coefficients[1]
    A = np.exp(B)

    print(f"A = {A}, n = {n}")
    
    row = {'Номер образца': material, 'A, 1 / [ сут * кг/см^2 ]': A, 'n': n} 
    
    df_result = df_result.append(row, ignore_index=True)
    
    sigma = []
    e = []
    for i in range(50,550):
        s = i*(10**4)/(10**7)
        sigma.append(s)
        e.append(A*(s**n))
        
    fig, ax = plt.subplots()
    ax.plot(np.array(sigma), np.array(e))
    plt.xlabel('sigma, кг/см^2')
    plt.ylabel('de/dt, 1 / сут')
    #plt.legend()
    plt.grid(True)
    plt.show()
```

```python
df_result
```

```python
merged_df = pd.merge(df_for_train, df_result, on='Номер образца', how='inner')
```

```python
merged_df
```

```python
merged_df.to_csv("A_n_results.csv")
```
